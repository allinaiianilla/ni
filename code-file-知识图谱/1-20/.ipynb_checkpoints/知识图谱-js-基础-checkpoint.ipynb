```typescript
// 对外宣布 一个自定义关键字 叫 声明
// 自定义一个关键字 叫 声明
// 同一个关键字 只能 声明 一次

let mood;
let age;
// 把值存入变量的操作成为赋值
mood = "happy js";
age = 331145; // 给变量 age 赋值 33

console.log(mood);
console.log(age);
```

```typescript
// 字符串由0个或多个字符构成
// 字符串必须被引号括起来，单引号或双引号都可以

let mood = "happy";
console.log(mood);
// let mood = 'bucuo'; // error：一个变量只能声明一次
mood = 'bucuo'
console.log(mood);

mood = 'don\'t ask'; 
// 通过 \ 表明字符串中间的单引号是普通单引号字符，而非字符串结束符号
// \ 被成为转义符
console.log(mood);
```

```typescript
// 创建一个数组
// 声明并赋值
let numbers = [1, 2, 3, 4, 5];
console.log(numbers);

// 数组的取数：names[i] 取索引对应的值
// 索引/下标（从0开始）
console.log(numbers[0]);

// 修改数组元素
console.log(numbers[0]);
numbers[0] = 0;
console.log(numbers[0]);

// 获取数组长度
console.log(numbers.length);

// 添加元素
numbers.push(6);
console.log(numbers);
```

```typescript
let person = {
    name: "Tom",
    age: 20,
    job: "developer"
};

console.log(person);
```

```typescript
// 访问对象属性
// 方法1：点号
console.log(person.name);

// 方法2：中括号
console.log(person["age"]);

// 修改对象
person.age = 21;
console.log(person.age);
// 添加新属性
console.log(person);
person.city = "Tokyo";
console.log(person);

// 删除属性
delete person.job;
console.log(person);

// 对象中也可以放函数(方法)
```

```typescript
/*
if (condition) {
    statements;
}
*/

if (2 > 1) {
    console.log("2 > 1 true");
}

if ( 2 > 3){
    console.log("2 > 3 true");
} else {
    console.log("2 > 3 false");
}
```

```typescript
// >, <, >=, <=, ===, !==

console.log(2 > 1);
console.log(2 < 1);
console.log(2 >= 1);
console.log(2 <= 1);
console.log(2 === 1);
console.log(2 !== 1);
```

```typescript
// 多个比较结果的组合结果
// && || !
let num = 5;

if (num >= 5 && num <= 10){
    console.log("The number is in the right range.");
} else {
    console.log("The number is not in the right range.");
}

if (num < 5 || num > 10) {
    console.log("The number is in the right range.");
} else {
    console.log("The number is not in the right range.");
}

if (!(num > 3)) {
    console.log("num <=3");
}else {
    console.log("num >3");
}
```

```typescript
// while循环
// 判断条件为真,执行代码块 进入下一轮循环
// 判断条件为假 停止执行代码块
// 语法格式
/*
while (condition) {
    statements;
}
*/

let n = 1;
while (n <= 5) {
    console.log(n);
    n = n + 1;
}
```

```typescript
// do-while循环
/*
do {
    statements;
} while (condition);
*/
let n = 1;
do {
    console.log(n);
    n = n + 1;
} while (n <= 5);
```

```typescript
// for循环
/*
for (initial condition; test condition; alter condition) {
    statements;
}
*/

for (let n=1; n <= 5; n = n+1) {
    console.log(n*2);
}

let names = ['zs', 'zy','ni','m']
for (let n=0; n < names.length; n = n+1) {
    console.log(names[n])
}
```

```typescript
// 当需要反复做一件事时,都可以利用函数来避免重复键入大量的相同内容

// 自定义函数/创建函数
function shout() {
    let names = ['zy','xm','ly'];
    for (let i=0; i<names.length;i++){
        console.log(names[i]);
    }
}

// 调用函数 call function
shout();

// 可以把数据传递给函数,函数处理该数据 或 根据该数据去完成预定的操作
// 传递给函数的数据称为 参数argument
/*
function name(arguments) {
    statements;
}
*/
function square(num) {
    let total = num * num;
    return total;
}

let nu = 9;
let tt = square(nu);

console.log(tt);
```
