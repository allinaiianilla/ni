```python
# 对一份数据去重——真实需求
# 有一份名单，大概200个名字，其中有部分重复，通过人工筛选很困难
# 程序就适合做这种 规则明确、持续完成的工作

names_str = """
1
2
1
3
"""
names = {name for name in names_str.split('\n')}
print(names)
```

```python
# 一句话的指令/代码/语句 一下就能看懂
# 200行代码 —— 已经完全蒙圈
# 写好注释是良好的编码习惯
# 功能说明、使用说明、需求说明 等等等等

# 该函数用于清洗数据
def clean_str(input_str):
    output_str = input_str.strip()
    output_str = output_str.lower()

    return output_str

# 调用上面功能代码（函数）
name = " Nijianghua "
print("清洗后的名字：",name)
name = clean_str(name)
print("清洗后的名字：",name)
```
