```python
import openai
API_KEY = "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"
model="meta/llama-3.1-8b-instruct"
def call_nvidia_api(prompt, model="meta/llama-3.1-8b-instruct", api_key="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"):
    client = openai.OpenAI(
        base_url="https://integrate.api.nvidia.com/v1",
        api_key=api_key
    )

    completion = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.2,
        top_p=0.7,
        max_tokens=10240,
        stream=True
    )

    for chunk in completion:
        # 修复 IndexError: 检查 choices 是否存在且非空
        if chunk.choices and len(chunk.choices) > 0:
            content = chunk.choices[0].delta.content
            if content is not None:
                print(content, end="")

# 使用示例
call_nvidia_api("你好，请自我介绍一下")
```
